#3. التعامل مع المشاكل

```python
import pandas as pd

# first lets drop the duplicates values,
df_clean = df.drop_duplicates()
# then lets also drop null values for both description and customerID
df_clean = df_clean.dropna(subset=['Description'])
df_clean = df_clean.dropna(subset=['CustomerID'])

print("Data Types Check (After Cleaning)")
df_clean.info()
#in a case of null or duplicate dropping the whole row is the best solution in this case where i can't fill the
# data myself and i can't work on empty data
```

```python
# To solve the non-sensical values in quantity and unit price (less or equal to 0)
# We will filter out those rows because these values are illogical for sales and would skew any analysis(avareg quantity or total sales)
df_clean = df_clean[(df_clean['Quantity'] > 0) & (df_clean['UnitPrice'] > 0)].copy()

# Correcting data types as to match thier description by converting them
# Enabling proper time-series analysis, allowing for insights into sales trends over time.
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])

# Improving memory efficiency and ensures the ID is treated as a discrete numerical identifier, and to do grouping or filtering operations
df_clean['CustomerID'] = df_clean['CustomerID'].astype(pd.Int64Dtype())
df_clean['Country'] = df_clean['Country'].astype('category')

print("check after treating datatypes and nonsensical values")
df_clean.info()
print(f"Current number of rows with Quantity <= 0: {df_clean[df_clean['Quantity'] <= 0].shape[0]}")
print(f"Current number of rows with UnitPrice <= 0: {df_clean[df_clean['UnitPrice'] <= 0].shape[0]}")


# While the problem of extreme values (outliers) still exist for Quantity and Unitprice
# We have chosen to not take action like capping or removing since removing all of them without understanding of the context might
# ene up eliminating actually valueable information like bulk orders or high-value orders

display(df_clean[['Quantity', 'UnitPrice']].describe())
print(f"1. **'Quantity'**: The maximum quantity is {df_clean['Quantity'].max()}")
print(f"2. **'UnitPrice'**: The maximum unit price is {df_clean['UnitPrice'].max()}")

# Update the main df for use in other cells
df = df_clean
```
